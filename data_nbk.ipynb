```python
%pip install -q pandas numpy
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/output/customers_education.csv")
df
```

```python
#age-group:
df["age_group"] = np.random.choice(["18-21","21-30","30-35","35-40","40-45","45-50","50-55","55-60"],
                                   size=df.shape[0],
                                   p=[0.40,0.24,0.2,0.1,0.03,0.01,0.01,0.01])
```

```python
df["age_group"].value_counts()
```

```python
##age
df["age"] = df["age_group"].apply(lambda x:np.random.randint(x.split('-')[0],x.split('-')[1])).astype('int')
```

```python
##gender
df["gender"] = np.random.choice(["M","F"],size=df.shape[0],p=[0.55,0.45])
```

```python
#education level
df["education_level"] = "None"
df.loc[df["age"]<=24,"education_level"] = "Bachelors"
df.loc[df["age"]<=21,"education_level"] = "High School Diploma"
df.loc[(df["age"]>24),"education_level"] = np.random.choice(["Bachelors","Masters"], size=df.loc[(df["age"]>24)].shape[0],
                                                            p=[0.6,0.4])
```

```python
#region
df["region"] = np.random.choice(["Urban","Rural"],df.shape[0],p=[0.55,0.45])
```

```python
#marital status
df["marital_status"] = np.random.choice(["Unmarried","Married"], size=df.shape[0],p=[0.2,0.8])
df.loc[(df["age"]<=24) & (df["gender"]=="M"),"marital_status"] = "Unmarried"
df.loc[(df["age"]<=21) & (df["gender"]=="F"),"marital_status"] = "Unmarried"
```

```python
df["marital_status"].value_counts()
```

```python
df["net_worth"] = 0
df.loc[(df["age"]<=20),"net_worth"] = np.random.randint(0,10,size=df.loc[(df["age"]<=20)].shape[0])*10000
df.loc[(df["age"]>20),"net_worth"] = np.random.randint(1,10,size=df.loc[(df["age"]>20)].shape[0])*50000
df.loc[(df["age"]>25),"net_worth"] = np.random.randint(3,201,size=df.loc[(df["age"]>25)].shape[0])*50000
```

```python
df["cibil"] = 0
df.loc[(df["age"]<=20),"cibil"] = np.random.randint(300,750,size=df.loc[(df["age"]<=20)].shape[0])
df.loc[(df["age"]>20),"cibil"] = np.random.randint(350,900,size=df.loc[(df["age"]>20)].shape[0])
df.loc[(df["net_worth"]==0),"cibil"] = 0
```

```python
df
```

```python
df["kyc_status"] = np.random.choice(["Compliant","Non-Compliant"],size=df.shape[0],p=[0.9,0.1])
```

```python
df = df.drop(columns="age_group",axis=1)
df
```

```python
df
```

```python
df1 = pd.read_csv("data/output/customers.csv")
df["date"] = df1["date"]
```

```python
df["id"] = df.index
```

```python
df.to_csv("data/output/customers_education.csv",index=False)
```
